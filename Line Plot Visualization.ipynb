Dataset: Immigration to Canada from 1980 to 2013 - [International migration flows to and from selected countries - The 2015 revision](https://www.un.org/development/desa/pd/data/international-migration-flows) from United Nation's website

In this lab, focus will be on the Canadian Immigration data and use the **already cleaned dataset** from previous lab 

```python
df_can = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/Canada.csv')

print('Data read into a pandas dataframe!')
```

```python
df_can.head()
```

The default index of the dataset is a numeric range from 0 to 194. This makes it very difficult to do a query by a specific country. For example to search for data on Haiti, there is need to know the corresponding index value.

This can be fixed by setting the 'Country' column as the index

```python
df_can.set_index('Country', inplace=True)
df_can.head(3)
```

```python
# removing the name of the index
df_can.index.name = None
df_can.head(3)
```

Declaring a variable that will allow easy call upon the full range of years:

```python
years = list(map(str, range(1980, 2014)))
```

## Line Plots

A line plot is a type of plot which displays information as a series of data points called 'markers' connected by straight line segments. It is when there is a continuous data set and best suited for trend-based visualizations of data over a period of time.

Creating a line plot to visualize the immigrants (to Canada) trend during 1980 to 2013. Usinge Total of year-wise immigrant,

```python
# creating df with only years columns from 1980 - 2013
df_line=df_can[years]

# Applying sum to get total immigrants year-wise
total_immigrants=df_line.sum()
total_immigrants
```

```python
# Creating figure and axes
fig, ax = plt.subplots()

# Plotting the line
ax.plot(total_immigrants)

#Setting up the Title
ax.set_title('Immigrants between 1980 to 2013') 
#Setting up the Labels
ax.set_xlabel('Years')
ax.set_ylabel('Total Immigrants')

plt.show()
```

The plot function populated the x-axis with the index values (years), and the y-axis with the column values (population). However, the years were not displayed because they are of type string. 

```python
# Changing the index values is to integer for plotting.
total_immigrants.index = total_immigrants.index.map(int)
```

```python
# Creating figure and axes
fig, ax = plt.subplots()

# Plotting the line
ax.plot(total_immigrants)

#Setting up the Title
ax.set_title('Immigrants between 1980 to 2013') 
#Setting up the Labels
ax.set_xlabel('Years')
ax.set_ylabel('Total Immigrants')

plt.show()
```

Customizing the above plot's appearance

```python
# Create figure and axes
fig, ax = plt.subplots()

# Customizing the appearance of Plot
ax.plot(total_immigrants, 
        marker='s', #Including markers in squares shapes
        markersize=5, #Setting the size of the marker
        color='green', #Changing the color of the line
        linestyle="dotted") #Changing the line style to a Dotted line
#Setting up the Title
ax.set_title('Immigrants between 1980 to 2013') 

#Setting up the Labels
ax.set_xlabel('Years')
ax.set_ylabel('Total Immigrants')
ax.legend(['Immigrants'])

plt.show()
```

Adding a background grid, a legend and changing the limits on the axis


```python
fig, ax = plt.subplots()

# Plotting the line
ax.plot(total_immigrants, 
        marker='s', #Including markers in squares shapes
        markersize=5, #Setting the size of the marker
        color='green', #Changing the color of the line
        linestyle="dotted") #Changing the line style to a Dotted line

# Setting up the Title
ax.set_title('Immigrants between 1980 to 2013') 

# Setting up the Labels
ax.set_xlabel('Years')
ax.set_ylabel('Total Immigrants')

# limits on x-axis
ax.set_xlim(1975, 2015)  #or plt.xlim()

# Enabling Grid
ax.grid(True)  #or plt.grid()

#Legend
ax.legend(["Immigrants"]) #or plt.legend()

# Display the plot
plt.show()
```

Using Haiti as a case study

In 2010, Haiti suffered a catastrophic magnitude 7.0 earthquake. The quake caused widespread devastation and loss of life and about three million people were affected by this natural disaster. As part of Canada's humanitarian effort, the Government of Canada stepped up its effort in accepting refugees from Haiti. This effort is visualized using a `Line` plot

```python
# changing the plot style
mpl.style.use('ggplot')
```

```python
# creating data series
haiti = df_can.loc['Haiti', years] 
haiti.head()
```

```python
haiti.plot()
```

```python
haiti.plot(kind='line')

plt.title('Immigration from Haiti')
plt.ylabel('Number of immigrants')
plt.xlabel('Years')

plt.show()
```

```python
# years are strings, changing the index values to integer.
haiti.index = haiti.index.map(int) 
haiti.plot(kind='line')

plt.title('Immigration from Haiti')
plt.ylabel('Number of Immigrants')
plt.xlabel('Years')

# syntax: plt.text(x, y, label)
plt.text(2000, 6300, '2010 Earthquake') 
plt.legend(["Haiti"])
plt.show() 
```

Since the Earthquake that happen in 2010, a lot of people emmigrated from Haiti and chose to go to Canada probably because the Government of Canada in her humanitarian effort welcomed more haitians.

Comparing the number of immigrants from India and China from 1980 to 2013.

```python
df_CI = df_can.loc[['India', 'China'], years]
df_CI
```

Since `df_CI` is a dataframe with the `country` as the index and `years` as the columns, the dataframe is transposed to swap the row and columns.

```python
df_CI = df_CI.transpose()
df_CI.head()
```

```python
# Changing the index values of df_CI to type integer for plotting
df_CI.index = df_CI.index.map(int) 

df_CI.plot(kind='line')

# Ysing scripting layer
plt.title('Immigrants from China and India')
plt.ylabel('Number of Immigrants')
plt.xlabel('Years')

plt.show()
```

From the above plot, it can be observed that the China and India have very similar immigration trends through the years. 

There is no need to transpose Haiti's dataframe before plotting (like df_CI). That's because `haiti` is a series as opposed to a dataframe, and has the years as its indices

Comparing the trend of top 5 countries that contributed the most to immigration to Canada.

```python
df_can1 = df_can.copy()
df_can1.sort_values(by='Total', ascending=False, axis=0, inplace=True)
df_top5 = df_can1.head(5)
df_top5
```

```python
df_top5 = df_top5[years].transpose()
df_top5
```

```python
# changing the index values of df_top5 to type integer for plotting
df_top5.index = df_top5.index.map(int) 

# pass a tuple (x, y) size
df_top5.plot(kind='line', figsize=(14, 8)) 

plt.title('Immigration Trend of Top 5 Countries')
plt.ylabel('Number of Immigrants')
plt.xlabel('Years')

plt.show()
```

The plot demonstrates diverse and fluctuating immigration trends among India, China, the United Kingdom, the Philippines, and Pakistan between 1980 and 2010, with the United Kingdom showing the highest numbers in certain periods, while India and China exhibit a general growth trend over time.
